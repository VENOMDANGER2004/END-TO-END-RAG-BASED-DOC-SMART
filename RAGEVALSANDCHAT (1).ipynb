```python
# --- STEP 1: Install Dependencies ---
# Run this cell first to install the necessary libraries for RAG and the Model
!pip install -q llama-index llama-index-llms-huggingface llama-index-embeddings-huggingface transformers accelerate bitsandbytes pypdf

# --- STEP 2: Import Libraries & Setup ---
import time
import torch
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.llms.huggingface import HuggingFaceLLM
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# Check if GPU is available (Crucial for speed)
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# --- STEP 3: Setup the Open Source Model (Phi-2) ---
# We use Microsoft's Phi-2 as suggested in your task (lightweight & fast)
# You can swap "microsoft/phi-2" with "mistralai/Mistral-7B-Instruct-v0.1" if you have Colab Pro.

print("Loading Phi-2 Model... (This may take 1-2 minutes)")

llm = HuggingFaceLLM(
    context_window=2048,
    max_new_tokens=512,
    generate_kwargs={"temperature": 0.1, "do_sample": False},
    tokenizer_name="microsoft/phi-2",
    model_name="microsoft/phi-2",
    device_map="auto",
    model_kwargs={"torch_dtype": torch.float16, "trust_remote_code": True}
)

# Set the embedding model (converts text to numbers for retrieval)
embed_model = HuggingFaceEmbedding(model_name="BAAI/bge-small-en-v1.5")

# Configure LlamaIndex to use our open-source setup
Settings.llm = llm
Settings.embed_model = embed_model
Settings.chunk_size = 512  # Smaller chunks often work better for specific contracts

print("Model and Embeddings Loaded Successfully!")

# --- STEP 4: Load Data & Create Index ---
# MAKE SURE you uploaded 'sample_contract.pdf' to the files tab on the left!
try:
    documents = SimpleDirectoryReader(input_files=["/content/sample_mortgage_document.pdf"]).load_data()
    index = VectorStoreIndex.from_documents(documents)
    query_engine = index.as_query_engine()
    print("Document Indexed Successfully!")
except FileNotFoundError:
    print("ERROR: Please upload 'sample_contract.pdf' to the Colab files area.")

# --- STEP 5: Run the Evaluation Queries ---
# These are the specific questions from your screenshots

queries = [
    "What are the penalties for late payments?",
    "Summarize the key terms in this contract.",
    "What is the refund policy?"
]

print("\n--- STARTING EVALUATION ---\n")

for query in queries:
    print(f"Query: {query}")

    # Measure Speed
    start_time = time.time()
    response = query_engine.query(query)
    end_time = time.time()

    duration = end_time - start_time

    print(f"Response (Phi-2): {response}")
    print(f"Time Taken: {duration:.2f} seconds")
    print("-" * 50)
```
