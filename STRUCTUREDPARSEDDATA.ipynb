```python
import fitz
import json
```

```python
print(f'PyMuPDF Ver:{fitz.__doc__}')
```

```python
filename='/content/LenderFeesWorksheetNew.pdf'
```

```python
doc=fitz.open(filename)
```

```python
structured_data=[]
```

```python
for page_num,page in enumerate(doc):
  words=page.get_text("words")
  for w in words:
    item={
        'text':w[4],
        'bbox':[w[0],w[1],w[2],w[3]],
        'page_num':page_num+1
    }
    structured_data.append(item)
```

```python
with open('mortgage_date.json','w') as f:
  json.dump(structured_data,f,indent=4)
```
