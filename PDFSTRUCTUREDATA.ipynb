```python
import fitz
import json

# 1. Load the PDF
filename = "LenderFeesWorksheetNew.pdf"
doc = fitz.open(filename)
page = doc[0] # The form is on Page 1
words = page.get_text("words")

# 2. Define the 6 Fields we want to find
# We map the "Label" found in the PDF to the "Field Name" we want in our list
target_fields = {
    "Amount:": "Loan Amount",
    "Rate:": "Interest Rate",
    "Program:": "Loan Program",
    "Applicant:": "Applicant Name",
    "Prepared:": "Date Prepared",
    "No:": "Application Number"
}

extracted_list = []

# 3. Logic to find the value NEXT to the label
for i, w in enumerate(words):
    text = w[4] # The text string

    # If the word matches one of our target labels...
    if text in target_fields:
        field_name = target_fields[text]

        # ... We take the NEXT word as the value (simple extraction)
        if i + 1 < len(words):
            val = words[i+1]

            # Create the entry with text AND bounding box (Required)
            entry = {
                "field_name": field_name,
                "extracted_value": val[4],
                "bbox": [val[0], val[1], val[2], val[3]]
            }
            extracted_list.append(entry)

# 4. Print the Final List (This is what you need to show)
print(json.dumps(extracted_list, indent=4))

# 5. Save to JSON (Optional, but good for records)
with open("key_fields.json", "w") as f:
    json.dump(extracted_list, f, indent=4)
```
