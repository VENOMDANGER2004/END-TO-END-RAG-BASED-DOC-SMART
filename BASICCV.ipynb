```python
IMGRGB=cv2.cvtColor(imgBGR,cv2.COLOR_BGR2RGB)
```

```python
# Apply Median Filtering
median = cv2.medianBlur(IMGRGB, 9)
```

```python
plt.subplot(1,2,1)
plt.imshow(IMGRGB)
plt.subplot(1,2,2)
plt.imshow(median)
plt.axis('off')
plt.show()
```

```python
clean_image=cv2.fastNlMeansDenoisingColored(median,None,10,10,7,21)
```

```python
clean_image
```

```python
kernel = np.array([[-1, -1, -1],
                   [-1,  9, -1],
                   [-1, -1, -1]])

sharpened_image = cv2.filter2D(clean_image, -1, kernel)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title("1. Original Noisy")
plt.imshow(IMGRGB)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("2. Median Blur (Cleaned)")
plt.imshow(clean_image)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("3. Sharpened (Final)")
plt.imshow(sharpened_image)
plt.axis('off')
```
