```python
import time
from google import genai
from google.genai.errors import ClientError

# --- SETUP ---
# 1. Initialize the new Client
client = genai.Client(api_key="AIzaSyAui9QaYaME9fFJztXHgSgAzV9Dd0mNCzE")

# 2. Start a chat session (using the stable 1.5 Flash model)
chat = client.chats.create(model="gemini-2.5-flash")

print("------------------------------------------------")
print("Chatbot started! (Type 'exit' to stop)")
print("------------------------------------------------")

# --- MAIN LOOP ---
while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        break

    try:
        # Send message using the new 'chat.send_message' syntax
        response = chat.send_message(user_input)
        print(f"AI: {response.text}")

    except ClientError as e:
        # Handle Rate Limits (429) or other API errors
        print(f"\n[!] API Error: {e}")
        print("[!] Waiting 30 seconds before retry...")
        time.sleep(30)

    except Exception as e:
        print(f"\n[Error] Something went wrong: {e}")
```
